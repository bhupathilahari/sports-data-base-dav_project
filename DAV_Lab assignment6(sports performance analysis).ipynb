### Load Dataset and Plot Boxplots
Let's load your dataset and explore its distribution using boxplots.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define the file path
file_path = '/content/final_dav_dataset_cleaned (2).csv.xls'

# Load the dataset (handling potential format differences due to the .xls extension on a CSV file)
try:
    # Try reading as CSV first
    df = pd.read_csv(file_path)
    print("Successfully loaded file as CSV.")
except Exception:
    try:
        # Fallback to Excel if CSV reading fails
        df = pd.read_excel(file_path)
        print("Successfully loaded file as Excel.")
    except Exception as e:
        print(f"Error loading file: {e}")

# Display basic information
if 'df' in locals():
    display(df.head())
    print(df.info())
```

```python
if 'df' in locals():
    # Set up a side-by-side plot comparing Attendees and Expenses grouped by Event Type
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # 1. Attendees grouped by Event Type
    sns.boxplot(data=df, x='Event Type', y='Attendees', ax=axes[0], hue='Event Type', legend=False, palette='Set2')
    axes[0].set_title('Attendees by Event Type')
    axes[0].grid(True, linestyle='--', alpha=0.5)

    # 2. Expenses grouped by Event Type
    sns.boxplot(data=df, x='Event Type', y='Expenses', ax=axes[1], hue='Event Type', legend=False, palette='Set2')
    axes[1].set_title('Expenses by Event Type')
    axes[1].grid(True, linestyle='--', alpha=0.5)

    plt.tight_layout()
    plt.show()
```

```python
if 'df' in locals():
    # Set up a side-by-side violin plot
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # 1. Violin plot for Attendees grouped by Event Type
    sns.violinplot(data=df, x='Event Type', y='Attendees', ax=axes[0], hue='Event Type', legend=False, palette='Set2', inner='quart')
    axes[0].set_title('Attendees Distribution by Event Type')
    axes[0].grid(True, linestyle='--', alpha=0.5)

    # 2. Violin plot for Expenses grouped by Event Type
    sns.violinplot(data=df, x='Event Type', y='Expenses', ax=axes[1], hue='Event Type', legend=False, palette='Set2', inner='quart')
    axes[1].set_title('Expenses Distribution by Event Type')
    axes[1].grid(True, linestyle='--', alpha=0.5)

    plt.tight_layout()
    plt.show()
```

```python
if 'df' in locals():
    # Set up side-by-side histograms
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # 1. Histogram for Attendees
    sns.histplot(data=df, x='Attendees', hue='Event Type', kde=True, multiple='stack', ax=axes[0], palette='Set2')
    axes[0].set_title('Distribution of Attendees')
    axes[0].grid(True, linestyle='--', alpha=0.5)

    # 2. Histogram for Expenses
    sns.histplot(data=df, x='Expenses', hue='Event Type', kde=True, multiple='stack', ax=axes[1], palette='Set2')
    axes[1].set_title('Distribution of Expenses')
    axes[1].grid(True, linestyle='--', alpha=0.5)

    plt.tight_layout()
    plt.show()
```
